```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog || '.' || 'divya_gold');
CREATE OR REPLACE VIEW IDENTIFIER(:catalog || '.' || 'gold_agg.top_5_customers') AS 
SELECT customer_id, SUM(total_amount) AS total_sales FROM IDENTIFIER(:catalog || '.' || 'divya_silver.sales_cleaned') GROUP BY customer_id ORDER BY total_sales DESC LIMIT 5
```

```python
%sql
create or replace view  IDENTIFIER(:catalog || '.' || 'divya_gold.top_5_products') as 
select product_id, round(sum(total_amount)) as total_sales from  IDENTIFIER(:catalog || '.' || 'divya_silver.sales_cleaned') group by product_id order by total_sales desc limit 5
```
